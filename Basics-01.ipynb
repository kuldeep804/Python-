### LIST
Lists are used to store multiple items in a single variable.
List items are ordered, changeable, and allow duplicate values.

```python
fruits = ["apple", "banana", "cherry", "apple", "cherry"]
print(fruits)
print(len(fruits))

# A list with strings, integers and boolean values:

list1 = ["abc", 34, True, 40, "male"]
print(list1)
print(len(list1))

#to make list using list() constructor
list2 = list(("apple", "banana", "cherry")) # note the double round-brackets
print(list2)

#List items are indexed and you can access them by referring to the index number:
fruits = ["apple", "banana", "cherry"]
print(fruits[1])
print(fruits[-1])
print(fruits[0:2]) # returns the items from index 0 to 2 (not included

#To change the value of a specific item, refer to the index number:
fruits = ["apple", "banana", "cherry"]
fruits[0] = "kiwi"
print(fruits)

#Insert Items
fruits = ["apple", "banana", "cherry"]
fruits.insert(1, "orange")
print(fruits)

#Extend List
fruits = ["apple", "banana", "cherry"]
more_fruits = ["mango", "pineapple", "papaya"]
fruits.extend(more_fruits)
print(fruits)

#Add Any Iterable
fruits = ["apple", "banana", "cherry"]
more_fruits = ("mango", "pineapple", "papaya") # note the parentheses
fruits.extend(more_fruits)
print(fruits)

#delete list items 
fruits = ["apple", "banana", "cherry"]
del fruits[0]
print(fruits)

#pop() method removes the specified index, (or the last item if index is not specified):
fruits = ["apple", "banana", "cherry"]
fruits.pop(1)
print(fruits)

#clear method empties the list: 
fruits = ["apple", "banana", "cherry"]
fruits.clear()
print(fruits)
```

### List Comprehension
List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

Example:

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
  if "a" in x:
    newlist.append(x)

print(newlist)

# The Syntax
# newlist = [expression for item in iterable if condition == True]


#With list comprehension you can do all that with only one line of code:
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x]
print(newlist)

#Expression
#The expression is the current item in the iteration, but it is also the outcome, which you can manipulate before it ends up like a list item in the new list:

newlist = [x.upper() for x in fruits if "a" in x]
print(newlist)

#short the list 
#alphabetical order
fruits.sort()
print(fruits)

#Sort Descending
fruits.sort(reverse = True)
print(fruits)


```
