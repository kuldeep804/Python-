## 1. Indentation

In Python, indentation is not just for readability; it defines the structure of the code. Blocks of code such as loops, conditions, and functions rely on consistent indentation.

> Python uses whitespace to decide what belongs to a block.

```python
## Indentation
## Python uses indentation to define blocks of code.
## Indentation is typically done using four spaces or a tab.
## Consistent indentation is crucial in Python, as it determines the structure and flow of the code.

age = 19
if age >= 18:
    print("You are an adult!")
else:
    print("You are a Minor!")
```

## 5. Lists

Lists are used to store multiple items in a single variable. List items are ordered, changeable, and can contain duplicate values.

### 5.1 Creating a list

```python
fruits = ["apple", "banana", "cherry", "apple", "cherry"]
print(fruits)
print(len(fruits))

# A list with strings, integers and boolean values:
list1 = ["abc", 34, True, 40, "male"]
print(list1)
print(len(list1))

# Create a list using the list() constructor
list2 = list(("apple", "banana", "cherry"))  # note the double round-brackets
print(list2)
```

### 5.2 Accessing and updating list items

```python
# List items are indexed and can be accessed by the index number:
fruits = ["apple", "banana", "cherry"]
print(fruits[1])
print(fruits[-1])
print(fruits[0:2])  # returns the items from index 0 to 2 (not included)

# Change the value of a specific item by index:
fruits = ["apple", "banana", "cherry"]
fruits[0] = "kiwi"
print(fruits)
```

### 5.3 Adding items

```python
# Insert items
fruits = ["apple", "banana", "cherry"]
fruits.insert(1, "orange")
print(fruits)

# Extend a list
fruits = ["apple", "banana", "cherry"]
more_fruits = ["mango", "pineapple", "papaya"]
fruits.extend(more_fruits)
print(fruits)

# Add any iterable
fruits = ["apple", "banana", "cherry"]
more_fruits = ("mango", "pineapple", "papaya")  # note the parentheses
fruits.extend(more_fruits)
print(fruits)
```

### 5.4 Removing items

```python
# Delete list items
fruits = ["apple", "banana", "cherry"]
del fruits[0]
print(fruits)

# pop() removes the specified index, or the last item if no index is provided
fruits = ["apple", "banana", "cherry"]
fruits.pop(1)
print(fruits)

# clear() empties the list
fruits = ["apple", "banana", "cherry"]
fruits.clear()
print(fruits)
```

## 6. List Comprehension

List comprehension provides a shorter and cleaner way to create a new list based on an existing one.

Example:

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
    if "a" in x:
        newlist.append(x)

print(newlist)

# Syntax:
# newlist = [expression for item in iterable if condition == True]

# With list comprehension, you can do the same in one line:
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x]
print(newlist)

# Expression: the current item can be transformed before being added to the list
newlist = [x.upper() for x in fruits if "a" in x]
print(newlist)

# Sort the list in alphabetical order
fruits.sort()
print(fruits)

# Sort descending
fruits.sort(reverse=True)
print(fruits)
```
